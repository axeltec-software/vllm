```python
import os
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd


data_val_mean_H = []
SCRIPT_DIR = "/home/imizus/projects/vllm_gonka_tests"
artifacts_dir = os.path.join(SCRIPT_DIR, "vllm_executor", "vllm", "artefacts_honest")
for dir in os.listdir(artifacts_dir):
    file = os.path.join(artifacts_dir, dir, "vllm_validator_top_p_probs_mean.txt")
    with open(file=file, mode='r') as f:
        lines = f.readlines()
    for l in lines:
        data_val_mean_H.append(float(l))

df_mean_H = pd.concat(axis=0, ignore_index=True, objs=[
    pd.DataFrame.from_dict({'value': np.array(data_val_mean_H).T, 'name': 'Mean'})
])

fig, ax = plt.subplots()

# Create the histogram
sns.histplot(data=df_mean_H, x='value', hue='name', multiple='dodge', ax=ax)
sns.move_legend(ax, bbox_to_anchor=(1, 0.5), loc='center left', frameon=False)

plt.title('Mean probability Distribution (Honest scenario)')
plt.xlabel('Probability')
plt.ylabel('Frequency')
plt.show()
```

```python
data_val_mean_F = []
artifacts_dir = os.path.join(SCRIPT_DIR, "vllm_executor", "vllm", "artefacts_fraud")
for dir in os.listdir(artifacts_dir):
    file = os.path.join(artifacts_dir, dir, "vllm_validator_top_p_probs_mean.txt")
    with open(file=file, mode='r') as f:
        lines = f.readlines()
    for l in lines:
        data_val_mean_F.append(float(l))

df_mean_F = pd.concat(axis=0, ignore_index=True, objs=[
    pd.DataFrame.from_dict({'value': np.array(data_val_mean_F).T, 'name': 'Mean'})
])

fig, ax = plt.subplots()

# Create the histogram
sns.histplot(data=df_mean_F, x='value', hue='name', multiple='dodge', ax=ax)
sns.move_legend(ax, bbox_to_anchor=(1, 0.5), loc='center left', frameon=False)

plt.title('Mean probability Distribution (Fraud scenario)')
plt.xlabel('Probability')
plt.ylabel('Frequency')
plt.show()
```

```python
print("Fraud mean distribution:")
print(f"Mean: {df_mean_F.loc[:, 'value'].mean()}")
print(f"Std: {df_mean_F.loc[:, 'value'].std()}")

```
