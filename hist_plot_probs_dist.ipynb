```python
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

data_val_max = []
with open(file='/home/imizus/projects/vllm/vllm_top_p_probs_max.txt', mode='r') as f:
    lines = f.readlines()
    for l in lines:
        data_val_max.append(float(l))

data_val_mean = []
with open(file='/home/imizus/projects/vllm/vllm_top_p_probs_mean.txt', mode='r') as f:
    lines = f.readlines()
    for l in lines:
        data_val_mean.append(float(l))

df = pd.concat(axis=0, ignore_index=True, objs=[
    pd.DataFrame.from_dict({'value': np.array(data_val_mean).T, 'name': 'Mean'})
])

fig, ax = plt.subplots()

# Create the histogram
sns.histplot(data=df, x='value', hue='name', multiple='dodge', ax=ax)
sns.move_legend(ax, bbox_to_anchor=(1, 0.5), loc='center left', frameon=False)

plt.title('Mean probability Distribution')
plt.xlabel('Probability')
plt.ylabel('Frequency')
plt.show()
```

```python
df = pd.concat(axis=0, ignore_index=True, objs=[
    pd.DataFrame.from_dict({'value': np.array(data_val_max).T, 'name': 'Max'})
])

fig, ax = plt.subplots()

# Create the histogram
sns.histplot(data=df, x='value', hue='name', multiple='dodge', ax=ax)
sns.move_legend(ax, bbox_to_anchor=(1, 0.5), loc='center left', frameon=False)

plt.title('Max probability Distribution')
plt.xlabel('Probability')
plt.ylabel('Frequency')
plt.show()
```
